```python
import mysql.connector

db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="college_records_db"
)

mycursor = db.cursor()

print("Database connected successfully!")
```

```python
sql = "INSERT INTO student_details (roll_no, student_name, score) VALUES (%s, %s, %s)"

value = (197, "Twinkal", 86)

mycursor.execute(sql, value)
db.commit()

print("Student record inserted successfully!")
```

```python
mycursor.execute("SELECT * FROM student_details")

result = mycursor.fetchall()

for record in result:
    print(record)
```

```python
sql = "UPDATE student_details SET score = %s WHERE roll_no = %s"

value = (94, 101)

mycursor.execute(sql, value)
db.commit()

print("Student record updated successfully!")
```

```python
mycursor.execute("SELECT * FROM student_details")

for record in mycursor.fetchall():
    print(record)
```

```python
sql = "DELETE FROM student_details WHERE roll_no = %s"

value = (101,)

mycursor.execute(sql, value)
db.commit()

print("Student record deleted successfully!")
```

```python
mycursor.execute("SELECT * FROM student_details")

result = mycursor.fetchall()

for record in result:
    print(record)
```

```python
sql = "INSERT INTO student_details (roll_no, student_name, score) VALUES (%s, %s, %s)"

records = [
    (101, "Krisha", 86),
    (102, "Vidhi", 91),
    (103, "Itisha", 89)
]

mycursor.executemany(sql, records)
db.commit()

print("Three student records inserted successfully!")
```

```python
mycursor.execute("SELECT * FROM student_details")

for record in mycursor.fetchall():
    print(record)
```
